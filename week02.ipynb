```python
#JAWABAN NOMOR 2

# KANVAS HITAM INT 16
black_int16 = np.zeros((500, 500, 3), dtype=np.int16)
# KANVAS HITAM INT 32
black_int32 = np.zeros((500, 500, 3), dtype=np.int32)

#informasi kanvas int16
print("=== INT16 ===")
print("dtype :", black_int16.dtype)
print("min   :", black_int16.min())
print("max   :", black_int16.max())
print("nbytes:", black_int16.nbytes)

print()

#informasi kanvas int32
print("=== INT32 ===")
print("dtype :", black_int32.dtype)
print("min   :", black_int32.min())
print("max   :", black_int32.max())
print("nbytes:", black_int32.nbytes)

# Menampilkan kedua kanvas
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(black_int16)
plt.title("Kanvas int16")

plt.subplot(1, 2, 2)
plt.imshow(black_int32)
plt.title("Kanvas int32")

plt.show()
```

Kanvas int16 dan int32 memiliki tampilan yang sama karena seluruh nilai pikselnya bernilai 0 sehingga keduanya berwarna hitam, dengan nilai min() dan max() yang juga sama-sama 0.
```python
# Membaca gambar
img = cv.imread('ktpsensor.jpg')

# Mengubah BGR menjadi RGB
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Melakukan pencerminan
flip_vertikal = cv.flip(img_rgb, 0)
flip_horizontal = cv.flip(img_rgb, 1)
flip_keduanya = cv.flip(img_rgb, -1)

# Menampilkan hasil
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertikal)
plt.title("Flip Vertikal")

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title("Flip Horizontal")

plt.subplot(1, 3, 3)
plt.imshow(flip_keduanya)
plt.title("Flip Keduanya")

plt.show()
```

### Penjelasan :

### Pertanyaan yang saya buat :
Bagaimana perubahan ukuran citra setelah dilakukan proses cv.resize() menjadi setengah dari ukuran aslinya?

```python
# JAWABAN :
# Membaca gambar
img = cv.imread('aktivitas.jpg')

# Menampilkan ukuran gambar asli
print("Ukuran gambar asli:", img.shape)

# Mengambil tinggi dan lebar gambar
h, w = img.shape[:2]

# Mengubah ukuran gambar menjadi setengah dari ukuran asli
img_resize = cv.resize(img, (w // 2, h // 2))

# Menampilkan ukuran gambar setelah resize
print("Ukuran gambar setelah resize:", img_resize.shape)

# Mengubah gambar dari BGR ke RGB
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
img_resize_rgb = cv.cvtColor(img_resize, cv.COLOR_BGR2RGB)

# Menampilkan gambar asli dan hasil resize
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Gambar Asli")

plt.subplot(1, 2, 2)
plt.imshow(img_resize_rgb)
plt.title("Gambar Setelah Resize")

plt.show()
```

### Rancangan Program

Program dimulai dengan membaca citra menggunakan cv.imread(), kemudian ukuran asli citra ditampilkan menggunakan img.shape. Selanjutnya, tinggi dan lebar citra diambil menggunakan img.shape[:2]. Ukuran citra kemudian diubah menjadi setengah dari ukuran sebelumnya menggunakan cv.resize(). Setelah itu, ukuran hasil resize ditampilkan dan kedua gambar dibandingkan menggunakan plt.imshow().

### data hasil :     
Ukuran gambar asli: (842, 1500, 3)

Ukuran gambar setelah resize: (421, 750, 3)

### Kesimpulan :    
Dari program yang telah saya buat, ukuran citra berubah setelah dilakukan proses cv.resize(). Tinggi dan lebar citra menjadi setengah dari ukuran citra asli sehingga jumlah pikselnya juga berkurang. Hal ini menunjukkan bahwa cv.resize() benar-benar mengubah ukuran data citra, berbeda dengan figsize yang hanya mengubah ukuran tampilan gambar tanpa mengubah nilai shape citra.

### Studi Kasus 2:
Marketplace seperti Tokopedia dan Shopee menampilkan foto
produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau
lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja.
Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail
persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna.
Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas,
tambahkan logo nama toko berukuran proporsional pada gambar.

```python
#JAWABAN :
# Membaca gambar produk
img = cv.imread('belalang.jpg')

# Mengambil tinggi dan lebar gambar
h, w = img.shape[:2]

# Menentukan ukuran thumbnail
size = 500

# Menentukan skala resize
scale = min(size / w, size / h)

# Menghitung ukuran baru
new_w = int(w * scale)
new_h = int(h * scale)

# Mengubah ukuran gambar
img_resize = cv.resize(img, (new_w, new_h))

# Membuat kanvas persegi
canvas = np.zeros((size, size, 3), dtype=np.uint8)

# Menentukan posisi gambar agar berada di tengah
x = (size - new_w) // 2
y = (size - new_h) // 2

# Menempatkan gambar pada kanvas
canvas[y:y + new_h, x:x + new_w] = img_resize

# Menambahkan nama brand
cv.putText(
    canvas,
    'ZK',
    (size // 20, size - size // 20),
    cv.FONT_HERSHEY_SIMPLEX,
    size / 300,
    (255, 255, 255),
    size // 100,
    cv.LINE_AA
)

# Mengubah BGR menjadi RGB
canvas_rgb = cv.cvtColor(canvas, cv.COLOR_BGR2RGB)

# Menampilkan hasil
plt.figure(figsize=(7, 7))
plt.imshow(canvas_rgb)
plt.title("Thumbnail Produk ZK")
plt.show()
```

### Penjelasan :
Program membuat thumbnail produk berbentuk persegi dengan rasio 1:1 menggunakan kanvas berukuran 500 × 500 piksel. Ukuran gambar produk dihitung terlebih dahulu untuk menentukan skala yang sesuai sehingga proses resize() tidak mengubah rasio asli gambar. Setelah ukurannya disesuaikan, gambar ditempatkan di tengah kanvas menggunakan perhitungan posisi berdasarkan ukuran gambar dan kanvas. Selanjutnya, nama brand ZK ditambahkan menggunakan cv.putText() sehingga menghasilkan thumbnail produk yang berbentuk persegi dan tetap mempertahankan proporsi gambar asli.

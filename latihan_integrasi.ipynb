```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
urls = ["https://i.pinimg.com/1200x/48/1e/80/481e80a5c7261d1acfa4001980708280.jpg",
        "https://i.pinimg.com/736x/d9/1a/ab/d91aab075550a621e371c4b6b77afe31.jpg",
        "https://i.pinimg.com/736x/16/3d/d0/163dd09e5a19081ed860f6999b50c8cc.jpg"]

for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.3, fy=0.3)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar", tinggi,"x",lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(tinggi):
  image_3[y, int(lebar / 2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(tinggi):

    x = int(y * (lebar / tinggi))

    image_3[y, x] = [255, 255, 255]

    # Diagonal 2: Kanan atas ke kiri bawah
    if (lebar - 1 - x) >= 0:
        image_3[y, lebar - 1 - x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

y_awal = int(tinggi / 3)
x_awal = int(lebar / 3)

# Menggunakan nested loop untuk mewarnai piksel bentuk kotak
for y in range(y_awal, y_awal + 100):
    for x in range(x_awal, x_awal + 100):
        # Proteksi biar gak error kalau kotaknya ngelebihi batas resolusi
        if y < tinggi and x < lebar:
            image_3[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
